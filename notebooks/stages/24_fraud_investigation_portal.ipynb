# Portal de investigação
Visão da janela → fila priorizada → caso com contexto, feedback e SHAP.
A API, Grafana e MLflow preservam suas responsabilidades. A interface consulta
um snapshot verificado das decisões HTTP aceitas, sem novo treino ou inferência.
Aceite nativo da captura e da experiência visual pendente.


## Comandos e guia
[Operação do portal](../../docs/operations/FRAUD_PORTAL.md).
Instalar: `make portal-install`. Preparar com digest SHAP e ausência de escritores;
verificar: `make portal-verify`; abrir: `make portal-open`; segundo terminal:
`make portal-smoke`. O frontend local atende em http://127.0.0.1:18501.
Todas as decisões do fluxo aparecem; cinco casos têm SHAP aceito.
