# Investigação dos casos do champion

Esta etapa complementa a fila e o contexto com as contribuições SHAP do HGB
que pontuou cada requisição HTTP. Os pesos e as features precisam corresponder
à decisão persistida; o relatório não altera o score, o modelo ou a triagem.

Aceite nativo do autor em 2026-10-22: cinco casos publicados e verificados
sem mudar o champion; verificação offline sem carregar modelo.


## Gerar e verificar

Após aplicar, revisar, testar e commitar o recorte, execute no terminal da raiz:

```bash
FRAUD_CASE_EXPLANATIONS_NO_OTHER_WRITERS=1 .venv/bin/python -m tox -e case-explanations
```

O ensaio reutiliza o fluxo aceito `champion-http-v1` e publica uma única versão
do relatório `champion-shap-v1`. Não execute também o gerador pelo Make para o
mesmo ID. Guarde o `manifest_sha256` impresso.

```bash
make product-explanations-verify PRODUCT_EXPLANATION_SHA256=DIGEST_IMPRESSO
make product-explanations PRODUCT_EXPLANATION_SHA256=DIGEST_IMPRESSO
xdg-open data/reports/case-explanations/champion-shap-v1/index.html
```


## Ler o resultado

O gráfico mostra contribuições em log-odds. Laranja aumenta e azul diminui a
margem; cinza indica zero. A base e a soma precisam recompor a margem do HGB,
cuja sigmoide reproduz o score HTTP. Ranking não é probabilidade calibrada.

O feedback posterior é contexto de investigação e não entra no SHAP. A seleção
por alertas/score não estima qualidade global. Uma explicação descreve o modelo,
não a causa da fraude. O Grafana mantém a observabilidade operacional.
