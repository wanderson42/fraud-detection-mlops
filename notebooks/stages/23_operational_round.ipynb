# Rodada operacional canônica

A unidade de demonstração é o ciclo completo, da origem dos dados ao modelo servido. A linhagem dos datasets e a decisão de promoção são requisitos de conclusão.


## Contrato antes da execução

O protocolo `operational_round_policy_v1.json` fixa os candidatos Dummy, Regressão Logística e HGB, as janelas Gold v1, AP mínima 0,5, precisão diária por cliente no top 100 mínima 0,45 e a necessidade de superar o Dummy em AP. O melhor candidato elegível aprovado recebe `champion`. Nenhum modelo histórico participa. A seleção usa validação; o teste histórico não é reaberto para escolher o vencedor.


## Execução e recursos

Reutilizamos os Jobs de dados e o servidor MLflow do KinD. O Job de treino baixa somente o Gold identificado por manifesto; apenas o init container recebe credenciais S3. O processo científico publica por HTTP e registra Inputs, schema, target, contextos, partições, hashes, parâmetros e ambiente. O Job usa até 2 CPUs e 4 GiB e termina após a rodada. O serving mantém uma API ativa e um gateway; o green só é ativado durante o ensaio.


## Aceite

O Registry só recebe uma versão após uma leitura independente dos datasets, dos três modelos e de suas métricas. O modelo selecionado é empacotado, testado no runtime atual e servido pelo gateway `fraud-detector`. O aceite final exige identidade do champion/HTTP, score de controle, HTTP 422 e troca/falha/rollback. O runbook contém os comandos executáveis; este caderno não executa novo treino ao ser aberto.
