# Evidência do modelo para investigação
A avaliação temporal é um relatório separado, ligado aos pesos/release do snapshot.
A UI mostra comparação, Precision–Recall, ROC e carga da política por cliente/dia.
O alvo binário não é um veredito sobre pagamentos; score, política, contexto e
feedback posterior têm papéis distintos. A interface não depende do algoritmo.
[Operação](../../docs/operations/PORTAL_PERFORMANCE.md).
`make portal-performance-prepare PRODUCT_NO_OTHER_WRITERS=1` →
`make portal-performance-verify` → reiniciar `make portal-open`.
Somente validação é pontuada, sem fit ou novo acesso ao holdout. Aceite nativo pendente.
