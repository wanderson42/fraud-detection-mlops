# 12 — Organização de dados e features

A segunda etapa tira ingestão, qualidade, contratos e construção de datasets da
árvore plana. O objetivo é permitir evolução da infraestrutura sem fazer a
modelagem depender dos executores de dados. Os contratos e algoritmos permanecem
os mesmos; esta entrega não inicia treinamento nem acessa holdouts reais.


## Decisão

Quatro áreas em `data/`: contratos, ingestão, datasets e qualidade. `features/`
separa schema, histórico causal e validação de valores. Mover tudo para um único
subdiretório conservaria a mistura de responsabilidades; criar componentes
futuros acrescentaria diretórios sem implementação. A
[arquitetura](../../docs/ARCHITECTURE.md#dados-e-features) mantém o mapa completo.

Na segunda etapa, paths antigos delegavam às implementações novas. Os adaptadores
foram retirados após a migração; imports e comandos atuais usam os módulos explícitos.
A segunda etapa é incremental sobre a primeira. Regras operacionais ficam no
[runbook](../../docs/operations/OPERATIONS.md#migração-de-dados-e-features).
