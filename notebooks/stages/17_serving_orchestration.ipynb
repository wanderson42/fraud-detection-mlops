O PR #3 integrou Docker em `main`. Este incremento adiciona Prefect 3.8.8 no host,

## Operação e recuperação

A operação inspeciona manifesto, modelo e controle sem carregar o estimator,
consulta health/readiness/identidade e pontua o controle por HTTP. O fluxo compõe
três tasks; só a consulta repete indisponibilidade transitória, até três tentativas.
Hashes, contrato ou score divergentes encerram a execução sem retry.

Um ID vincula release, URL e código. Recibo completo existente implica `reused: true`
sem nova consulta. Use outro ID para verificar o serviço novamente. Publicação local
atômica evita dois recibos do mesmo ID; consultas podem repetir sob concorrência.


## Executar no Alienware

Siga os três terminais do [procedimento](../../docs/operations/SERVING_CONTRACT.md#iniciar-os-três-processos-locais):
servidor Prefect em 4200, Docker existente em 8000 e CLI do fluxo. A execução nativa
com a referência real ainda precisa ser relatada. Não execute comandos deste notebook
para reexportar modelos, abrir janelas ou retreinar.

Conferir o recibo e repetir o mesmo ID demonstra retomada. Registrar sucesso do fluxo
não transforma a pontuação de controle em teste de carga ou SLA.
