# Sucessores e modelo ativo

Rodada inicial aceita: HGB versão 1, Inputs de treino/validação, API e rollback verificados pelo autor. O novo protocolo compara candidato e incumbent, preserva champion durante a preparação e exige aprovação vinculada ao digest para alterar tráfego.

Busca por modelo: fixa, grid ou Optuna; busca desligada por padrão. Tuning recebe somente janelas internas de treino. Comparação usa rótulos maduros e os mesmos dados para todos.

Política inicial: ganho AP ≥ 0,005, queda P@100 ≤ 0,01 e mínimos de qualidade. Igualdade de desempenho mantém o champion. A validação já foi examinada; este é um ensaio operacional, sem alegação de generalização independente.

Implementação e testes locais concluídos; ensaio nativo da substituição pendente. O primeiro comando é `make update-validate`; execute depois o runbook. Este caderno não inicia treino nem autoriza mudança de tráfego.
