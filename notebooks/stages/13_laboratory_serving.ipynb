## Decisão e responsabilidade

O primeiro serviço recebe features prontas. Históricos causais, feedback e fila
de investigação pertencem a capacidades seguintes. O contrato executável usa
os tipos e coerências da Gold; a ordem no JSON não altera a matriz do modelo.

O exportador offline usa a avaliação salva e o tracking existente, preserva bytes
skops e compara os scores da validação após ida e volta por JSON. Não reabre Gold
de treino/teste, não ajusta e não publica outra run. Não reescreve o freeze antigo
para fazer seus hashes de código coincidir com a organização nova.


## Distribuição e carga

A release contém modelo, caso de controle sem rótulos e manifesto identificado.
O SHA-256 do manifesto deve ser fixado pelo operador. Na inicialização, o serviço
verifica hashes, tipos revisados, features, ambiente e score de controle. O modelo
carrega uma vez; requisições não consultam MLflow.

`main` reúne dependências de inferência; `pipeline`, ferramentas offline; `dev`,
checks. A instalação padrão continua completa e preserva versões do lockfile.
O wheel HTTP foi instalado num ambiente novo sem a stack de experimentação e
respondido por Uvicorn fora do checkout. Protocolos dos workflows offline continuam
recursos externos: este check comprova o runtime de serving.
